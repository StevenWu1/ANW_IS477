```python
import pandas as pd
import numpy as np
```

```python
# import data
bus = pd.read_csv("../Data/Bus_Routes.csv", low_memory = False)
l_station = pd.read_csv("../Data/L_Stations.csv")
temp = pd.read_csv("../Data/Temperature.csv")
water = pd.read_csv("../Data/Water.csv")
```

## Bus Data Exploration

```python
bus.head()
```

```python
# Describe every column
for col in bus.columns:
    print(col)
    print(bus[col].describe())
    print()
```

```python
# Check for any NA
for col in bus.columns:
    nulls = sum(bus[col].isna())
    print(col, "has", nulls, "nulls")
```
