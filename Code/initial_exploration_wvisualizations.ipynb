## Data Visualization

```python
# Cleaning for Analyis
# Convert dates to datetime
bus["date"] = pd.to_datetime(bus["date"])
l_station["date"] = pd.to_datetime(l_station["date"])
temp["DATE"] = pd.to_datetime(temp["DATE"])
water["DATE"] = pd.to_datetime(water["DATE"])

# Convert ridership to numeric
bus["rides"] = pd.to_numeric(
    bus["rides"].astype(str).str.replace(",", ""),
    errors="coerce"
)

l_station["rides"] = pd.to_numeric(
    l_station["rides"].astype(str).str.replace(",", ""),
    errors="coerce"
)
```

```python
# Visualization 1: Monthly Bus Ridership
monthly_bus = bus.groupby(bus["date"].dt.to_period("M"))["rides"].sum()
monthly_bus.plot(
    kind="line",
    figsize=(12, 5),
    title="Monthly Chicago Bus Ridership")

plt.xlabel("Date")
plt.ylabel("Total Rides")
plt.show()
```

```python
# Visualization 2: Most Popular Routes
top_routes = bus.groupby("route")["rides"].sum().nlargest(10)
top_routes.sort_values().plot(
    kind="barh",
    figsize=(9, 5),
    title="Top 10 Bus Routes by Total Ridership")

plt.xlabel("Total Rides")
plt.ylabel("Bus Route")
plt.show()
```

```python
# Visualization 3: Average Ridership by Day
day_rides = bus.groupby("daytype")["rides"].mean()

day_rides.plot(
    kind="bar",
    title="Average Bus Ridership by Day Type"
)

plt.xlabel("Day Type")
plt.ylabel("Average Rides per Route-Day")
plt.show()
```

```python
# Visualization 4: Temperature vs. Ridership
daily_bus = bus.groupby("date")["rides"].sum().reset_index()
daily_temp = temp.groupby("DATE")["TMAX"].mean().reset_index()

weather_bus = daily_bus.merge(
    daily_temp,
    left_on="date",
    right_on="DATE")

weather_bus.plot(
    kind="scatter",
    x="TMAX",
    y="rides",
    figsize=(9, 5),
    title="Maximum Temperature vs. Daily Bus Ridership")

plt.xlabel("Maximum Temperature (°F)")
plt.ylabel("Daily Bus Rides")
plt.show()
```
